```python
import struct, os, glob

parts = sorted(glob.glob('./galaxy_inputs/input/barcodrain_resources.7z.*'))
total = sum(os.path.getsize(p) for p in parts)

with open(parts[0], 'rb') as f:
    h = f.read(32)
offset, size = struct.unpack('<QQ', h[12:28])
expected = 32 + offset + size

print(f"Taille réelle    : {total:,}")
print(f"Taille attendue  : {expected:,}")
print(f"Différence       : {expected - total:,}")
```
