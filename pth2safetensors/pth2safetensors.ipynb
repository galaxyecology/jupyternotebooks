```python
import json
import torch
from safetensors.torch import save_file

# 1. Charger le fichier .pth
pth_path = "galaxy_inputs/pth/DeepForestVisionV2.pth.zip"
data = torch.load(pth_path, map_location="cpu", weights_only=True)

# Gérer si le .pth contient l'objet complet ou directement le state_dict
state_dict = data.get("state_dict", data) if isinstance(data, dict) and "state_dict" in data else data
if not isinstance(state_dict, dict):
    state_dict = data.state_dict()

# 2. Sauvegarder les tenseurs au format safetensors
safetensors_path = "outputs/collection/model.safetensors"
# S'assurer que les tenseurs sont continus (contiguous)
tensors = {k: v.contiguous() for k, v in state_dict.items() if isinstance(v, torch.Tensor)}
save_file(tensors, safetensors_path)
print(f"Poids convertis avec succès vers : {safetensors_path}")

# 3. Créer un fichier config.json de base (adaptez selon votre architecture)
config_data = {
    "model_type": "custom",
    "format": "safetensors",
    "num_parameters": sum(p.numel() for p in tensors.values())
}

config_path = "outputs/collection/config.json"
with open(config_path, "w", encoding="utf-8") as f:
    json.dump(config_data, f, indent=4)
print(f"Configuration JSON créée avec succès vers : {config_path}")
```
