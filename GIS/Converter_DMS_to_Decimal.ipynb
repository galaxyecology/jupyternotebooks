```python
import os
import pandas as pd
from lat_lon_parser import parse

df = pd.read_csv("galaxy_inputs/herbier/UC1-3c-open-refine.csv.txt", sep=";", encoding="utf-8")

def convert_coord(val):
    if pd.isna(val):
        return None
    val = str(val).strip()
    try:
        # déjà en décimal
        return float(val)
    except ValueError:
        # sinon on essaie de parser en DMS
        return parse(val)

df['lat'] = df['lat'].apply(convert_coord)
df['lon'] = df['lon'].apply(convert_coord)

df[['lat', 'lon']].head(30)
df = df.fillna("NA")

print(df[['lat', 'lon']])

df.to_csv("./outputs/UC1-3c-open-refine_jupyter.tsv", sep="\t", index=False, encoding="utf-8")

```
